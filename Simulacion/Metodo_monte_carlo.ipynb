```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
 return np.sin(x)
def monte_carlo(a, b, n):
    m = np.linspace(a, b, 500)
    ys = np.max(f(m))
    yi = np.min(f(m))
    y = np.random.uniform(0, 1, n)
    x = np.random.uniform(a, b, n)
    exitos = []
    fracasos = []
    count = 0
    for i in range(n):
        if abs(y[i]) <= abs(f(x[i])):
            count += 1
            exitos.append([y[i], x[i]])
        else:
            fracasos.append([y[i], x[i]])
    c = count /  n
    c = c*((b - a)*ys)
    print("Valor aproximado de la integral: ", c)
    return exitos, fracasos
e, f = monte_carlo(0, np.pi/2, 100)
plt.plot(np.linspace(0, np.pi/2, 100), np.sin(np.linspace(0, np.pi/2, 100)))
k, j = zip(*e)
plt.scatter(j, k)
        
```
